```python
# ================== INSTALL ==================
!pip install -q transformers torch

# ================== IMPORTS ==================
from transformers import pipeline, set_seed, GPT2LMHeadModel, GPT2Tokenizer
import torch

# ================== SET SEED ==================
set_seed(42)

# ================== PIPELINE ==================
generator = pipeline(
    'text-generation',
    model='gpt2',
    device=0 if torch.cuda.is_available() else -1
)

prompt = "Once upon a time in a distant galaxy,"

outputs = generator(
    prompt,
    max_new_tokens=100,
    num_return_sequences=3,
    temperature=0.8,
    top_p=0.95,
    do_sample=True,
    pad_token_id=50256
)

print("=== Text Generations ===")
for i, output in enumerate(outputs):
    print(f"\nGeneration {i+1}:\n{output['generated_text']}")
    print("-" * 80)

# ================== MANUAL ==================
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')

tokenizer.pad_token = tokenizer.eos_token

prompt2 = "In a world where AI has become conscious,"

inputs = tokenizer(prompt2, return_tensors="pt").to(
    0 if torch.cuda.is_available() else "cpu"
)

outputs2 = model.generate(
    **inputs,
    max_new_tokens=120,
    temperature=0.7,
    top_p=0.9,
    do_sample=True,
    repetition_penalty=1.2
)

print("\n=== Manual Generation ===")
print(tokenizer.decode(outputs2[0], skip_special_tokens=True))
```
